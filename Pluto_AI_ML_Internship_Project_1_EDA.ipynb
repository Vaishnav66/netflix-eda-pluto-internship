## Data Cleaning

The following data cleaning steps were performed:

1. A copy of the original dataset was created so that the original data remained unchanged.

2. Missing values in the `director`, `cast`, `country`, and `rating` columns were replaced with `"Unknown"`. This preserves the records while clearly indicating that the original information was unavailable.

3. The `date_added` column was converted to datetime format. The remaining missing dates were handled using a placeholder value because the original dates were unavailable.

4. Missing values in the `duration` column were replaced with `"Unknown"` because the actual duration was not available and should not be guessed.

5. Duplicate rows were checked and removed if any were present.

These cleaning steps were performed to improve data quality while preserving as much of the original information as possible.

# Exploratory Data Analysis

Exploratory Data Analysis (EDA) is performed to understand the patterns, trends, distributions, and relationships present in the Netflix dataset. In this section, different questions are analyzed using Pandas and visualizations.

## Question 1: What is the distribution of Movies and TV Shows?

We first analyze the `type` column to determine how the Netflix content is distributed between Movies and TV Shows.

````python
content_type = df_clean["type"].value_counts()

content_type
````

````python
content_percentage = df_clean["type"].value_counts(normalize=True) * 100

content_percentage
````

````python
plt.figure(figsize=(8, 5))

plt.bar(content_type.index, content_type.values)

plt.title("Distribution of Movies and TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
````

## Question 3: Which countries contribute the most Netflix titles?

The `country` column may contain multiple countries for a single title. Therefore, the country values are split so that each country can be counted individually. This provides a more meaningful comparison of country-wise contributions to the dataset.

````python
country_data = df_clean["country"].str.split(", ").explode()

country_data.value_counts().head(10)
````

````python
df_clean["country"]
````

````python
top_countries = country_data.value_counts().head(10)

top_countries
````

````python
plt.figure(figsize=(10, 6))

plt.barh(top_countries.index, top_countries.values)

plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
````

### Finding

The analysis identifies the countries with the highest number of titles in the Netflix dataset. The chart shows that the distribution of Netflix titles is concentrated among a relatively small number of countries, with the countries at the top contributing the largest number of titles.

## Question 4: What are the most common genres on Netflix?

The `listed_in` column contains the genres or categories associated with each title. Since a title can belong to multiple genres, the values are split and counted individually to identify the most common genres in the dataset.

````python
genre_data = df_clean["listed_in"].str.split(", ").explode()

genre_data.value_counts().head(10)
````

````python
top_genres = genre_data.value_counts().head(10)

top_genres
````

````python
plt.figure(figsize=(10, 6))

plt.barh(top_genres.index, top_genres.values)

plt.title("Top 10 Genres on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
````

### Finding

The analysis shows that Netflix contains a wide variety of genres. The top 10 genres indicate which categories occur most frequently in the dataset, with the highest-ranked genres appearing across a large number of titles.

## Question 5: What is the distribution of movie durations?

The `duration` column contains the duration of Movies in minutes and the number of seasons for TV Shows. Therefore, only Movies are considered for analyzing movie duration.

````python
movie_data = df_clean[df_clean["type"] == "Movie"].copy()

movie_data["duration_minutes"] = movie_data["duration"].str.extract(r"(\d+)").astype(float)

movie_duration = movie_data["duration_minutes"].dropna()

movie_duration.describe()
````

````python
plt.figure(figsize=(10, 6))

plt.hist(movie_duration, bins=30, edgecolor="black")

plt.title("Distribution of Movie Durations on Netflix")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")

plt.show()
````

## Additional Visualization 1: Content Type Distribution

A pie chart is used to show the percentage share of Movies and TV Shows in the Netflix dataset.

````python
plt.figure(figsize=(7, 7))

plt.pie(
    content_type.values,
    labels=content_type.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage Distribution of Movies and TV Shows")

plt.show()
````

## Additional Visualization 2: Movie Duration vs Release Year

A scatter plot is used to examine the relationship between movie release year and movie duration.

````python
scatter_data = movie_data.dropna(subset=["duration_minutes"])

plt.figure(figsize=(10, 6))

plt.scatter(
    scatter_data["release_year"],
    scatter_data["duration_minutes"],
    alpha=0.3
)

plt.title("Movie Duration vs Release Year")
plt.xlabel("Release Year")
plt.ylabel("Duration (minutes)")

plt.show()
````

## Additional Visualization 3: Correlation Heatmap

A correlation heatmap is used to visualize the relationship between numerical variables in the dataset.

````python
# Create a correlation matrix
correlation_data = scatter_data[["release_year", "duration_minutes"]]

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap of Movie Data")

plt.show()
````
